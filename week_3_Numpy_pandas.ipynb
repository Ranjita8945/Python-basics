## creating numpy array

```python
numbers = np.array([10,20,30,40,50,60])

print(numbers)
print(type(numbers))
```

## numpy array operations

```python
numbers = np.array([10, 20, 30, 40, 50])

print("Addition:", numbers + 5)
print("Subtraction:", numbers - 5)
print("Multiplication:", numbers * 2)
print("Division:", numbers / 10)
```

## numpy aggregtion operations

```python
numbers = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(numbers))
print("Mean:", np.mean(numbers))
print("Maximum:", np.max(numbers))
print("Minimum:", np.min(numbers))
```

## numpy broadcasting

```python
numbers = np.array([10, 20, 30, 40, 50])

result = numbers + 10

print("Original array:", numbers)
print("After adding 10:", result)
```

## 2D numpy array

```python
marks = np.array([
    [80, 75, 90],
    [65, 88, 70],
    [92, 85, 95]
])

print(marks)
print("Shape:", marks.shape)
```

## Access Elements in a 2D NumPy Array

```python
marks = np.array([
    [80, 75, 90],
    [65, 88, 70],
    [92, 85, 95]
])

print("First row:", marks[0])
print("First value:", marks[0, 0])
print("Second row, third value:", marks[1, 2])
```

## NumPy Array Calculations

```python
marks = np.array([
    [80, 75, 90],
    [65, 88, 70],
    [92, 85, 95]
])

print("Total of all marks:", np.sum(marks))
print("Average of all marks:", np.mean(marks))

print("Total for each student:", np.sum(marks, axis=1))
print("Average for each student:", np.mean(marks, axis=1))
```

## numpy filtering

```python
marks = np.array([45, 67, 82, 39, 91, 56, 73])

passed_marks = marks[marks >= 50]

print("All marks:", marks)
print("Passed marks:", passed_marks)
```

## Pandas Series

```python
import pandas as pd

marks = pd.Series([80, 75, 90, 65, 88])

print(marks)
print("Average:", marks.mean())
print("Maximum:", marks.max())
print("Minimum:", marks.min())
```

## Pandas Series with Custom Index

```python
marks = pd.Series(
    [80, 75, 90, 65, 88],
    index=["Math", "Python", "SQL", "ML", "Statistics"]
)

print(marks)
```

```python
#access a particular value
print("Python marks:", marks["Python"])
print("ML marks:", marks["ML"])
```

## Creating a Pandas DataFrame

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul"],
    "Age": [20, 21, 20, 22],
    "Marks": [85, 78, 92, 74]
}

df = pd.DataFrame(data)

print(df)
```

## Accessing DataFrame Columns

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul"],
    "Age": [20, 21, 20, 22],
    "Marks": [85, 78, 92, 74]
}

df = pd.DataFrame(data)

print("Names:")
print(df[["Name", "Marks"]])

```

## Pandas Indexing with loc and iloc

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul"],
    "Age": [20, 21, 20, 22],
    "Marks": [85, 78, 92, 74]
}

df = pd.DataFrame(data)

print("First row:")
print(df.iloc[0])

print("\nFirst two rows:")
print(df.iloc[0:2])
```

## Pandas loc

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul"],
    "Age": [20, 21, 20, 22],
    "Marks": [85, 78, 92, 74]
}

df = pd.DataFrame(data)

print("First row:")
print(df.loc[0])

print("\nName and Marks:")
print(df.loc[:, ["Name", "Marks"]])
```

## Filtering Data in Pandas

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul"],
    "Age": [20, 21, 20, 22],
    "Marks": [85, 78, 92, 74]
}

df = pd.DataFrame(data)

high_marks = df[df["Marks"] >= 80]

print(high_marks)
```

## Add a New Column to a DataFrame

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul"],
    "Age": [20, 21, 20, 22],
    "Marks": [85, 78, 92, 74]
}

df = pd.DataFrame(data)

df["Passed"] = df["Marks"] >= 50

print(df)
```

## Modify DataFrame Values

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul"],
    "Age": [20, 21, 20, 22],
    "Marks": [85, 78, 92, 74]
}

df = pd.DataFrame(data)

df.loc[1, "Marks"] = 82

print(df)
```

## Check Basic DataFrame Information

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows:")
print(df.head())
```

## Descriptive Statistics with Pandas

```python
print("Average marks:", df["Marks"].mean())
print("Highest marks:", df["Marks"].max())
print("Lowest marks:", df["Marks"].min())
print("Total marks:", df["Marks"].sum())
```

## Grouping Data with groupby()

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul", "Sneha"],
    "Branch": ["AIML", "CSE", "AIML", "CSE", "AIML"],
    "Marks": [85, 78, 92, 74, 88]
}

df = pd.DataFrame(data)

branch_average = df.groupby("Branch")["Marks"].mean()

print(branch_average)
```

## Grouping with Multiple Aggregations

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul", "Sneha"],
    "Branch": ["AIML", "CSE", "AIML", "CSE", "AIML"],
    "Marks": [85, 78, 92, 74, 88]
}

df = pd.DataFrame(data)

branch_stats = df.groupby("Branch")["Marks"].agg(
    ["mean", "max", "min", "count"]
)

print(branch_stats)
```

## Handling Missing Values

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul", "Sneha"],
    "Marks": [85, 78, None, 74, 88],
    "Age": [20, 21, 20, None, 22]
}

df = pd.DataFrame(data)

print("Original data:")
print(df)

print("\nMissing values:")
print(df.isnull().sum())
```

## Remove Missing Values

```python
cleaned_df = df.dropna()

print("Original number of rows:", len(df))
print("Rows after removing missing values:", len(cleaned_df))

print("\nCleaned data:")
print(cleaned_df)
```

## Fill Missing Values

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul", "Sneha"],
    "Marks": [85, 78, None, 74, 88],
    "Age": [20, 21, 20, None, 22]
}

df = pd.DataFrame(data)

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df)
```

## Sorting Data with Pandas

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul", "Sneha"],
    "Marks": [85, 78, 92, 74, 88]
}

df = pd.DataFrame(data)

sorted_df = df.sort_values("Marks", ascending=True)

print(sorted_df)
```

## Save a DataFrame as a CSV File

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul"],
    "Marks": [85, 78, 92, 74]
}

df = pd.DataFrame(data)

df.to_csv("student_marks.csv", index=False)

print("Dataset saved successfully!")
```

## Practice value_counts()

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul", "Sneha", "Amit"],
    "Branch": ["AIML", "CSE", "AIML", "CSE", "AIML", "CSE"],
    "Marks": [85, 78, 92, 74, 88, 81]
}

df = pd.DataFrame(data)

print(df["Branch"].value_counts())
```

## Filtering with Multiple Conditions

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul", "Sneha", "Amit"],
    "Branch": ["AIML", "CSE", "AIML", "CSE", "AIML", "CSE"],
    "Marks": [85, 78, 92, 74, 88, 81]
}

df = pd.DataFrame(data)

result = df[(df["Branch"] == "AIML") & (df["Marks"] >= 85)]

print(result)
```

## Combining NumPy and Pandas

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul"],
    "Marks": [85, 78, 92, 74]
}

df = pd.DataFrame(data)

df["Marks_Doubled"] = np.array(df["Marks"]) * 2

print(df)
```

## Rename Columns

```python
data = {
    "student_name": ["Ranjita", "Anu", "Priya", "Rahul"],
    "student_marks": [85, 78, 92, 74]
}

df = pd.DataFrame(data)

df = df.rename(columns={
    "student_name": "Name",
    "student_marks": "Marks"
})

print(df)
```

## Remove Duplicate Rows

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Anu", "Rahul"],
    "Marks": [85, 78, 92, 78, 74]
}

df = pd.DataFrame(data)

print("Original data:")
print(df)

print("\nNumber of duplicate rows:", df.duplicated().sum())

df_cleaned = df.drop_duplicates()

print("\nAfter removing duplicates:")
print(df_cleaned)
```

## Check for Missing Values After Cleaning

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul"],
    "Marks": [85, 78, 92, 74],
    "Age": [20, 21, None, 22]
}

df = pd.DataFrame(data)

print("Missing values before cleaning:")
print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())

print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nCleaned DataFrame:")
print(df)
```

## Week 3 Client Project: Start the Dataset

```python
data = {
    "Name": ["Ranjita", "Anu", "Priya", "Rahul", "Sneha", "Amit", "Kiran", "Neha"],
    "Branch": ["AIML", "CSE", "AIML", "CSE", "AIML", "CSE", "AIML", "CSE"],
    "Maths": [85, 78, 92, 74, None, 81, 88, 76],
    "Python": [90, 82, 95, 70, 86, None, 91, 79],
    "Attendance": [92, 85, 96, 78, 88, 91, None, 83]
}

df = pd.DataFrame(data)

print(df)
```

## Inspect the Client Dataset

```python
print("First 5 rows:")
print(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

## Find Missing Values

```python
print(df.isnull().sum())
```

## Finding Missing Values

```python
df["Maths"] = df["Maths"].fillna(df["Maths"].mean())
df["Python"] = df["Python"].fillna(df["Python"].mean())
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean())

print(df)
```

```python
print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

## Check for Duplicate Rows

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
cleaned_df = df.drop_duplicates()

print("Rows before removing duplicates:", len(df))
print("Rows after removing duplicates:", len(cleaned_df))
```

## Calculate Average Marks

```python
print("Average Maths marks:", df["Maths"].mean())
print("Average Python marks:", df["Python"].mean())
print("Average Attendance:", df["Attendance"].mean())
```

## Calculate Average by Branch

```python
branch_average = df.groupby("Branch")[["Maths", "Python", "Attendance"]].mean()

print(branch_average)
```

## Find the Highest-Performing Branch

```python
df["Overall_Average"] = df[["Maths", "Python"]].mean(axis=1)

branch_performance = df.groupby("Branch")["Overall_Average"].mean()

print(branch_performance)
```

## Save the Cleaned Client Dataset

```python
cleaned_df = df.drop_duplicates()

cleaned_df.to_csv("student_performance_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
