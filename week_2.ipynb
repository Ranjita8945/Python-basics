## 5. Functions

```python
#Basic function
def greet():
    print("Hello! Welcome to Python.")

greet()
```

```python
#Function with parameters
def greet_user(name):
    print("Hello", name)

greet_user("Ranjita")
greet_user("Anu")
```

```python
#Function that returns a value
def add_numbers(a, b):
    return a + b

result = add_numbers(10, 20)

print("Result:", result)
```

```python
#Function for data transformation
def square_numbers(numbers):
    result = []

    for number in numbers:
        result.append(number ** 2)

    return result


numbers = [1, 2, 3, 4, 5]

squared = square_numbers(numbers)

print("Original:", numbers)
print("Squared:", squared)
```

```python
#Function to filter data
def get_even_numbers(numbers):
    even_numbers = []

    for number in numbers:
        if number % 2 == 0:
            even_numbers.append(number)

    return even_numbers


numbers = [1, 2, 3, 4, 5, 6]

print(get_even_numbers(numbers))
```

## 6. Lambda functions

```python
#Basic lambda
square = lambda x: x ** 2

print(square(5))
```

```python
#Lambda with two values
add = lambda a, b: a + b

print(add(10, 20))
```

```python
#Using lambda to transform data
numbers = [1, 2, 3, 4, 5]

squared = list(map(lambda x: x ** 2, numbers))

print(squared)
```

```python
#Lambda for filtering
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

## 7. Recursion

```python
def countdown(n):
    if n == 0:
        print("Done!")
        return

    print(n)
    countdown(n - 1)


countdown(5)
```

```python
#Factorial using recursion
def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)


print("Factorial:", factorial(5))
```

## 8. List comprehensions

```python
numbers = [1, 2, 3, 4, 5]

squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)
```

```python
numbers = [1, 2, 3, 4, 5]

squares = [number ** 2 for number in numbers]

print(squares)
```

```python
# Creating a list of numbers
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = [number for number in numbers if number % 2 == 0]

print(even_numbers)
```

```python
#Convert words to uppercase
names = ["ranjita", "anu", "priya"]

uppercase_names = [name.upper() for name in names]

print(uppercase_names)
```

```python
#Practical data transformation
prices = [100, 200, 300, 400]

discounted_prices = [price * 0.9 for price in prices]

print(discounted_prices)
```

## Hands-on data transformation

```python
#sum of squares
numbers = [1, 2, 3, 4, 5]

def sum_of_squares(numbers):
    total = 0

    for number in numbers:
        total += number ** 2

    return total


result = sum_of_squares(numbers)

print("Numbers:", numbers)
print("Sum of squares:", result)
```

```python
#Filtering data
def filter_greater_than_50(numbers):
    filtered = []

    for number in numbers:
        if number > 50:
            filtered.append(number)

    return filtered


numbers = [10, 25, 55, 70, 35, 90]

result = filter_greater_than_50(numbers)

print("Original data:", numbers)
print("Filtered data:", result)
```

```python
#The same filtering using list comprehension
numbers = [10, 25, 55, 70, 35, 90]

filtered = [number for number in numbers if number > 50]

print(filtered)
```

## Week 2 Client Project: Data Cleaning Script

###  10. Client Project: Data Cleaning Script

This program demonstrates basic data cleaning by:
- Removing duplicate values
- Removing invalid values
- Filtering the data

```python
# Data Cleaning Script

data = [10, 20, 20, 30, -5, 40, 40, 50, -10, 60]

print("Original data:", data)

# Step 1: Remove duplicate values
unique_data = list(set(data))

print("After removing duplicates:", unique_data)

# Step 2: Remove invalid negative values
cleaned_data = [value for value in unique_data if value >= 0]

print("After removing invalid values:", cleaned_data)

# Step 3: Calculate basic statistics
print("Number of valid values:", len(cleaned_data))
print("Minimum value:", min(cleaned_data))
print("Maximum value:", max(cleaned_data))
print("Average value:", sum(cleaned_data) / len(cleaned_data))
```
